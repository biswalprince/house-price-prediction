```python
DATA_PATH = '/kaggle/input/competitions/house-prices-advanced-regression-techniques/'

train = pd.read_csv(DATA_PATH + 'train.csv')
test = pd.read_csv(DATA_PATH + 'test.csv')

print("Datasets loaded succesfully!")

print("Training data:", train.shape)
print("Testing data:", test.shape)
```

```python
train_processed = train.copy()
test_processed = test.copy()
```

## 3. Missing Value Investigation

### 3.1 Garage Features

We investigate missing values across garage-related features to understand whether they indicate houses without garages.

```python
garage_columns = [
    'GarageType',
    'GarageFinish',
    'GarageQual',
    'GarageCond',
    'GarageYrBlt',
    'GarageCars',
    'GarageArea'
]
```

```python
display(
    train_processed[garage_columns]
    [train_processed['GarageType'].isnull()]
    .head(10)
)
```

```python
no_garage = train_processed[
    train_processed['GarageType'].isnull()
]

print("Number of houses without GarageType:", len(no_garage))
print("\nGarageCars values:")
print(no_garage['GarageCars'].value_counts(dropna=False))
```

### 3.2 Handling Missing Garage Values

Our investigation showed that houses with missing GarageType also have zero garage capacity.

We interpret these missing values as houses without garages.

Categorical garage features will be filled with "None", while GarageYrBlt will be filled with 0.

```python
# Categorical garage columns
garage_categorical = [
    'GarageType',
    'GarageFinish',
    'GarageQual',
    'GarageCond'
]

# Fill categorical missing values
train_processed[garage_categorical] = (
    train_processed[garage_categorical].fillna('None')
)

test_processed[garage_categorical] = (
    test_processed[garage_categorical].fillna('None')
)

# Fill missing garage construction year
train_processed['GarageYrBlt'] = (
    train_processed['GarageYrBlt'].fillna(0)
)

test_processed['GarageYrBlt'] = (
    test_processed['GarageYrBlt'].fillna(0)
)

print("Garage cleaning executed!")
```

```python
print("Training data:")
print(train_processed[garage_categorical + ['GarageYrBlt']].isnull().sum())

print("\nTesting data:")
print(test_processed[garage_categorical + ['GarageYrBlt']].isnull().sum())
```

```python
basement_columns = [
    'BsmtQual',
    'BsmtCond',
    'BsmtExposure',
    'BsmtFinType1',
    'BsmtFinType2',
    'TotalBsmtSF',
    'BsmtFullBath',
    'BsmtHalfBath'
]

print("Training data:")
print(train_processed[basement_columns].isnull().sum())

print("\nTesting data:")
print(test_processed[basement_columns].isnull().sum())
```

```python
print("TRAINING — missing rows")

display(
    train_processed[train_processed.isnull().any(axis=1)]
    [['Id', 'Electrical']]
)

print("\nTESTING — missing rows")

for col in test_processed.columns:
    missing_rows = test_processed[test_processed[col].isnull()]

    if not missing_rows.empty:
        print(f"\n{col}:")
        display(missing_rows[['Id', col]])
```

```python
categorical_missing = [
    'Electrical',
    'MSZoning',
    'Utilities',
    'Exterior1st',
    'Exterior2nd',
    'KitchenQual',
    'Functional',
    'SaleType'
]

for col in categorical_missing:
    print(col, "→", train_processed[col].mode()[0])
```

```python
# Fill categorical missing values using training modes
for col in categorical_missing:
    mode_value = train_processed[col].mode()[0]

    train_processed[col] = train_processed[col].fillna(mode_value)
    test_processed[col] = test_processed[col].fillna(mode_value)


# Fill structural numerical missing values with zero
test_processed['BsmtHalfBath'] = test_processed['BsmtHalfBath'].fillna(0)

test_processed['GarageCars'] = test_processed['GarageCars'].fillna(0)
test_processed['GarageArea'] = test_processed['GarageArea'].fillna(0)
```
