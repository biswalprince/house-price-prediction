```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)

print("Libraries imported succesfully!")
```

```python
DATA_PATH = '/kaggle/input/competitions/house-prices-advanced-regression-techniques/'

train = pd.read_csv(DATA_PATH + 'train.csv')
test = pd.read_csv(DATA_PATH + 'test.csv')

print("Datasets loaded succesfully")
```

```python
print("Training dataset shape:", train.shape)
print("Testing dataset shape:", test.shape)
```

```python
# Display the first five words
train.head()
```

```python
# Display last five records
train.tail()
```

```python
# Information about the dataset
train.info()
```

```python
# Statistical summary of numerical columns
train.describe()
```

```python
# Separate numerical and categorical columns

numerical_features = train.select_dtypes(
    include=['int64', 'float64']
).columns

categorical_features = train.select_dtypes(
    include=['object']
).columns

print("Number of numerical features:", len(numerical_features))
print("Number of categorical fetaures:", len(categorical_features))

print("\nNumerical features:")
print(list(numerical_features))

print("\nCategorical features:")
print(list(categorical_features))
```

```python
all_missing = train.isnull().sum()
filtered_missing = all_missing[all_missing > 0]
sorted_missing = filtered_missing.sort_values(
    ascending=False
)
print(sorted_missing)
```
