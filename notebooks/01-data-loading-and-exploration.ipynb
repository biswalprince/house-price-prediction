```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)

print("Libraries imported succesfully!")
```

```python
DATA_PATH = '/kaggle/input/competitions/house-prices-advanced-regression-techniques/'

train = pd.read_csv(DATA_PATH + 'train.csv')
test = pd.read_csv(DATA_PATH + 'test.csv')

print("Datasets loaded succesfully")
```

```python
print("Training dataset shape:", train.shape)
print("Testing dataset shape:", test.shape)
```

```python
# Display last five records
train.tail()
```

```python
# Information about the dataset
train.info()
```

```python
# Statistical summary of numerical columns
train.describe()
```

```python
# Separate numerical and categorical columns

numerical_features = train.select_dtypes(
    include=['int64', 'float64']
).columns

categorical_features = train.select_dtypes(
    include=['object']
).columns

print("Number of numerical features:", len(numerical_features))
print("Number of categorical fetaures:", len(categorical_features))

print("\nNumerical features:")
print(list(numerical_features))

print("\nCategorical features:")
print(list(categorical_features))
```

```python
all_missing = train.isnull().sum()
filtered_missing = all_missing[all_missing > 0]
sorted_missing = filtered_missing.sort_values(
    ascending=False
)
print(sorted_missing)
```

```python
train['SalePrice'].describe()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(train['SalePrice'], kde=True)

plt.title('Distribution of House Sale Prices')
plt.xlabel('Sale Price')
plt.ylabel('Frequency')

plt.show()
```

### 3.1 Target Variable Analysis

The target variable is SalePrice. We analyse its distribution and statistical properties to understand house prices in the dataset.

```python
train['SalePrice'].describe()
```

```python
print("Mean:", train['SalePrice'].mean())
print("Median:", train['SalePrice'].median())
print("Standard Deviation:", train['SalePrice'].std())
```

```python
plt.figure(figsize=(10, 4))

sns.boxplot(x=train['SalePrice'])

plt.title('Distribution of House Sale Prices')
plt.xlabel('Sale Price')

plt.show()
```

### 3.2 Numerical Feature Analysis

We explore relationships between numerical features and SalePrice to understand which characteristics influence house prices.

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=train,
    x='GrLivArea',
    y='SalePrice'
)

plt.title('Living Area vs Sale Price')
plt.xlabel('Living Area (sq ft)')
plt.ylabel('Sale Price')

plt.show()
```

### 3.3 Correlation Analysis

Correlation helps us identify numerical features that have strong relationships with our target variable, SalePrice.

```python
correlation = train.corr(numeric_only=True)

saleprice_corr = correlation['SalePrice'].sort_values(
    ascending=False
)

print(saleprice_corr.head(15))
```

```python
plt.figure(figsize=(10, 8))

top_features = saleprice_corr.head(10).index

sns.heatmap(
    train[top_features].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap - Top 10 Features')
plt.show()
```

### 3.4 Categorical Feature Analysis

Categorical features describe different characteristics of houses.
We analyse their distributions and relationships with SalePrice.

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    data=train,
    x='BldgType',
    order=train['BldgType'].value_counts().index
)

plt.title('Distribution of Building Types')
plt.xlabel('Building Type')
plt.ylabel('Number of Houses')

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    data=train,
    x='BldgType',
    y='SalePrice'
)

plt.title('Building Type vs Sale Price')
plt.xlabel('Building Type')
plt.ylabel('Sale Price')

plt.show()
```

### 3.5 Neighbourhood Analysis

We investigate how house prices vary across different neighbourhoods.

```python
neighborhood_prices = (
    train.groupby('Neighborhood')['SalePrice']
    .median()
    .sort_values(ascending=True)
)

plt.figure(figsize=(10, 9))

plt.barh(
    y=neighborhood_prices.index,
    width=neighborhood_prices.values
)

plt.title('Median House Price by Neighbourhood')
plt.xlabel('Median Sale Price')
plt.ylabel('Neighbourhood')

plt.tight_layout()
plt.show()
```


## 4. EDA Conclusions

### Key Findings

4. GarageCars, GarageArea and TotalBsmtSF also show positive correlations with SalePrice.
5. Different building types have different sale-price distributions.
6. Neighbourhood has a noticeable relationship with house prices.
