## 2. Separating Features and Target

We separate the target variable (SalePrice) from the input features.

- X contains the independent variables.
- y contains the target variable (SalePrice).

```python
X = train.drop('SalePrice', axis=1)
y = train['SalePrice']

X_test = test.copy()

print("Training features:", X.shape)
print("target:", y.shape)
print("Testing features:", X_test.shape)
```

## 3. Categorical Encoding

Machine learning models generally require numerical input.

We identify categorical features and convert them into numerical representations using suitable encoding techniques.

```python
categorical_features = X.select_dtypes(
    include=['object']
).columns

numerical_features = X.select_dtypes(
    exclude=['object']
).columns

print("Categorical features:", len(categorical_features))
print("Numerical features:", len(numerical_features))

print("\nCategorical columns:")
print(list(categorical_features))
```

```python
print("Number of neighborhoods:", X['Neighborhood'].nunique())

print("\nNeighborhood categories:")
print(sorted(X['Neighborhood'].unique()))
```

```python
neighborhood_encoded = pd.get_dummies(
    X['Neighborhood'],
    prefix='Neighborhood',
    dtype=int
)

display(neighborhood_encoded.head())
```

```python
print("Original columns:", X['Neighborhood'].shape[0])
print("Encoded shape:", neighborhood_encoded.shape)
```

```python
print(
    "Unique values in row sums:",
    neighborhood_encoded.sum(axis=1).unique()
)
```
